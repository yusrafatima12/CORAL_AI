```python
from collections import Counter
from itertools import islice
import matplotlib.pyplot as plt
from datasets import load_dataset

# Dataset load karo, sirf class_map column lo (images nahi)
ds = load_dataset("reefsupport/marine-data", "coral-masks",
                  split="train", streaming=True)
ds_small = ds.select_columns(["class_map"])

# Pehli 1000 images mein har class ginti karo
n = 1000
counts = Counter()
for s in islice(ds_small, n):
    names = {c["label_native"] for c in s["class_map"]}
    counts.update(names)

# Print karo
print("Total images dekhi:", n)
for name, c in counts.most_common():
    print(f"{name}: {c} images")

# Bar chart
names = [k for k, _ in counts.most_common()]
vals = [v for _, v in counts.most_common()]
plt.figure(figsize=(8, 6))
plt.barh(names[::-1], vals[::-1])
plt.xlabel(f"Kitni images mein (total {n})")
plt.title("Dataset mein har class kitni images mein hai")
plt.tight_layout()
plt.show()
```

```python
from itertools import islice
import matplotlib.pyplot as plt
from datasets import load_dataset

ds = load_dataset("reefsupport/marine-data", "coral-masks",
                  split="train", streaming=True)
ds_small = ds.select_columns(["class_map"])

def health_group(name):
    if "bleached" in name:
        return "bleached"
    if "dead" in name:
        return "dead"
    if "alive" in name:
        return "alive"
    return None   # baqi sab (sand, fish, sponge waghera)

n = 1000
group_counts = {"alive": 0, "bleached": 0, "dead": 0}

for s in islice(ds_small, n):
    groups_in_image = set()
    for c in s["class_map"]:
        g = health_group(c["label_native"])
        if g:
            groups_in_image.add(g)
    for g in groups_in_image:
        group_counts[g] += 1

print("Total images dekhi:", n)
for g, c in group_counts.items():
    print(f"{g}: {c} images ({100*c/n:.1f}%)")

plt.figure(figsize=(6, 4))
plt.bar(group_counts.keys(), group_counts.values(),
        color=["green", "orange", "gray"])
plt.ylabel(f"Images (total {n})")
plt.title("Coral health groups in dataset")
plt.show()
```
